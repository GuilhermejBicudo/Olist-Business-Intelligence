### Customer Dataset Quality Assessment

The `customers` table contains 99,441 records and 5 columns.

The dataset does not contain missing values, meaning all customer attributes are populated.

The analysis identified:

- `customer_id` as a candidate primary key, since it contains 99,441 unique values matching the total number of records.
- `customer_unique_id` as a customer-level identifier, containing 96,096 unique values, indicating that some customers have multiple records in the dataset.

This distinction is important because `customer_id` represents the customer record associated with an order, while `customer_unique_id` represents the actual customer entity.

### Key Findings

- The `customers` table contains 99,441 records and no missing values.
- `customer_id` can be considered a candidate primary key.
- `customer_unique_id` should not be treated as a primary key because multiple records can belong to the same customer.
- The relationship between customers and orders is expected to be one-to-many (`customers` → `orders`).

### Relationship with Orders

The `customers` table is expected to have a one-to-many relationship with the `orders` table.

Relationship:

customers.customer_id → orders.customer_id

One customer can be associated with multiple orders.

# Data Model Overview

After exploring the available datasets, the main entities, relationships, and table granularities were identified.

This overview represents the logical structure of the dataset before data transformation and analytical modeling.

## Dataset Structure

| Table | Grain | Primary Key / Candidate Key | Main Relationships |
|---|---|---|---|
| orders | 1 row = 1 order | order_id | customer_id, order_items, order_payments, order_reviews |
| customers | 1 row = 1 customer record | customer_id | orders |
| order_items | 1 row = 1 product item within an order | order_id + order_item_id | orders, products, sellers |
| products | 1 row = 1 product | product_id | order_items |
| sellers | 1 row = 1 seller | seller_id | order_items |
| order_payments | 1 row = 1 payment transaction | order_id + payment_sequential | orders |
| order_reviews | 1 row = 1 customer review event | review_id (not unique) | orders |
| geolocation | 1 row = 1 postal code location | zip_code_prefix (candidate) | customers, sellers |
| product_category_name_translation | 1 row = 1 category translation | product_category_name | products |

## Logical Relationships


                         customers
                             |
                             |
                         customer_id
                             |
                             ↓
                         orders
                             |
              +--------------+--------------+
              |              |              |
              ↓              ↓              ↓
        order_items    order_payments  order_reviews
              |
        +-----+------+
        |            |
        ↓            ↓
    products      sellers

## Key Modeling Observations

- The `orders` table represents the central entity of the business process, containing one record per order.
- The `order_items` table represents the transactional level, containing the products purchased in each order.
- The `customers`, `products`, and `sellers` tables represent descriptive entities that can be modeled as dimensions.
- The `order_payments` and `order_reviews` tables represent events associated with orders and may contain multiple records per order.
- The dataset structure suggests a dimensional model approach for analytical purposes, with `order_items` as the main fact table.
